## PuLP Solver Testi

Küçük bir LP modeli çözüyoruz: `max x + y`  
Kısıt: `x + y <= 1`, `x, y >= 0`  
Beklenen sonuç: `x = 0` veya `x = 1` (herhangi bir uç nokta), optimal değer **1.0**

```python
import pulp

prob = pulp.LpProblem("test", pulp.LpMaximize)
x = pulp.LpVariable("x", lowBound=0)
y = pulp.LpVariable("y", lowBound=0)

prob += x + y
prob += x + y <= 1

prob.solve(pulp.PULP_CBC_CMD(msg=0))

status = pulp.LpStatus[prob.status]
obj_val = pulp.value(prob.objective)

print(f"Durum  : {status}")
print(f"Optimal değer: {obj_val}")

solver_ok = (status == "Optimal" and abs(obj_val - 1.0) < 1e-6)
if solver_ok:
    print("\n✅  Solver çalışıyor")
else:
    print("\n❌  Solver beklenen sonucu vermedi — kurulumu kontrol edin")
    all_ok = False
```
